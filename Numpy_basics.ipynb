-------------Lists-------------

```python
grades = [85,90,78,92,88]
print("\n==Python list==")
print("list:",grades)
print("First Element",grades[0])#index start at 0
print("Last Element",grades[-1])#index last at -1
```

-----------Loops----------

```python
print("\n loop prints each grade")
for g in grades :
  print(g,end="")
  print()
```

--- List comprehension (concise loop) ---

```python
double = [g*2 for g in grades]
print("\n=== List comprehension: grades × 2 ===")
print("Original:",grades)
print("Double:",double)
```

```python
print("=== TRAP: list * 2 REPEATS — it does NOT multiply! ===")
print("grades*2=",grades*2)
```
