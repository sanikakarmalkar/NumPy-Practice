```python
import numpy as np
```

```python
arr = np.array([10,20,30,40,50,60])
print("Array",arr)
```

```python
## --- Positive indexing (left to right, starts at 0) ---
print("\n ===Positive Indexing===")
print("arr[0]",arr[0])
print("arr[2]",arr[2])
print("arr[5]",arr[5])
```

```python
## --- Negative indexing (right to left, starts at -1) ---
print("/n # --- Negative indexing (right to left, starts at -1) ---")
print("arr[-1]",arr[-1])
print("arr[-2]",arr[-2])
print("arr[-6]",arr[-6])
```

```python
# --- Slicing: arr[start : stop : step] ---
# stop is EXCLUSIVE — arr[1:4] gives indices 1, 2, 3 (NOT 4)

print("\n === slicing ===")
print("arr[1:4]",arr[1:4])
print("arr[:3]",arr[:3])
print("arr[3:]",arr[3:])
print("arr [::2]",arr[::2])
print("arr [1::2]",arr[1::2])
print("arr[::-1]",arr[::-1])
```

```python
# --- View vs Copy ---
print("\n===View vs Copy===")
original = np.array([1,2,3,4,5])
view_arr = original[1:4]
copy_arr = original[1:4].copy()

view_arr[0]=999
print("After view_arr[0]=999:")
print("Original=",original)

copy_arr[0]=888
print("After copy arr[0]=888:")
print("Original=",original)

```
